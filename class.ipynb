```python
class Student:

  def __init__(self, name="", roll=0, marks=0.0):
    self.name = name
    self.roll = roll
    self.marks = marks


# Initialize student instance
s1 = Student()

# Assign values
s1.name = "Rahul"
s1.roll = 101
s1.marks = 87.5

# Print details
print(f"Name:{s1.name}")
print(f"Roll:{s1.roll}")
print(f"Marks:{s1.marks:.1f}")
```
